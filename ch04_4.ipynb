```python
import cv2
from google.colab.patches import cv2_imshow

# 이미지 불러오기 및 RGB 변환
img = cv2.imread('/content/gdrive/My Drive/ColabIP26/Images/lena.jpg')
RGB_Img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# 그레이스케일로 변환
gImg1 = cv2.cvtColor(RGB_Img, cv2.COLOR_RGB2GRAY)

# 변환된 그레이 이미지 출력
cv2_imshow(gImg1)
```

```python
hist = cv2.calcHist([gImg1], [0], None, [256], [0, 256])
plt.plot(hist)
plt.show()
```

```python
gOut = cv2.equalizeHist(gImg1)
cv2_imshow(gOut)
```

```python
hist2 = cv2.calcHist([gOut], [0], None, [256], [0, 256])
plt.plot(hist2)
plt.show()
```

```python
cumHist = hist.copy()
for i in range(1, len(cumHist)):
  cumHist[i] = cumHist[i-1] + hist2[i]

plt.plot(cumHist)
plt.show()
```
