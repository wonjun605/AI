```python
import cv2
from google.colab.patches import cv2_imshow

# 이미지 불러오기 및 RGB 변환
Img = cv2.imread('/content/gdrive/My Drive/ColabIP26/Images/lena256.jpg')
RGB_Img = cv2.cvtColor(Img, cv2.COLOR_BGR2RGB)

# 그레이스케일로 변환
gImg= cv2.cvtColor(RGB_Img, cv2.COLOR_RGB2GRAY)
print(Img.shape)
cv2_imshow(gImg)
```

```python
from random import randint
```

```python
noiseImg = gImg.copy()
noiseRatio = 20;

for h in range(0, gImg.shape[0]):
  for w in range(0, gImg.shape[1]):
    val = randint(0, 100)
    if(val < noiseRatio/2):
      noiseImg[h, w] = 255
    elif (val < noiseRatio):
      noiseImg[h, w] = 0

cv2_imshow(noiseImg)
```

```python
outImg = cv2.medianBlur(noiseImg, 3)
cv2_imshow(outImg)
```
